### **Questions:**
1. How does weight vary across obesity levels?  
2. How are height and weight related for each obesity level? 
3. Which obesity levels occur most often in the dataset?
4. What does the age distribution look like?
5. Does physical activity differ across obesity levels?


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
ROOT_DIR = Path.cwd()
data_path = (ROOT_DIR.parent/"data"/"clean_data"/"cleaned_obesity_data.csv").resolve()
df = pd.read_csv(data_path)
```

#### **Q1:** How does weight vary across obesity levels? 

```python
plt.figure(figsize=(16, 8))

sns.boxplot(
    data=df,
    x="obesity_level",
    y="Weight",
    hue="obesity_level",
    palette="Set2",
    legend=False
)

plt.title("Weight Distribution Across Obesity Levels")
plt.xlabel("Obesity Level")
plt.ylabel("Weight (kg)")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

#### **Q2:** How are height and weight related for each obesity level? 

```python
plt.figure(figsize=(12,8))
sns.scatterplot(
    data=df,
    x="Height",
    y="Weight",
    hue="obesity_level", 
    palette="Set1", 
    alpha=0.7,
    s=50 
)
plt.title("Height vs Weight for Each Obesity Level")
plt.xlabel("Height (m)")
plt.ylabel("Weight (kg)")
plt.legend(bbox_to_anchor=(1.01, 1.01), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# height, weight, gender, obesity level relationship
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Height',
    y='Weight',
    hue='obesity_level',
    style='Gender',  
    palette='Set2',
    alpha=0.85,
    s=65,
    edgecolor='black',
    linewidth=0.7
)

plt.title('Relationship Between Height, Weight, and Obesity Level by Gender')
plt.xlabel('Height (m)')
plt.ylabel('Weight (kg)')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

#### **Q3:** Which obesity levels occur most often in the dataset?

```python
plt.figure(figsize=(16,6))
sns.catplot(
    data=df,
    y="obesity_level",
    kind="count",
    order=df["obesity_level"].value_counts().index,
    color="darkorange",
    width=0.5
)
plt.title("Frequency of Obesity Levels")
plt.show()
```
